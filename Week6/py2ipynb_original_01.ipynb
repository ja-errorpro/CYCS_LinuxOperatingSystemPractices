```
import nbformat
from nbformat.v4 import new_notebook, new_code_cell
import os
import sys

def convert_py_to_ipynb( py_path, ipynb_path = None ) :
  with open( py_path, 'r', encoding='utf-8' ) as f :
    lines = f.readlines()
  
  cells = []
  buffer = []
  
  for line in lines :
    if line.strip() == '# %%' :
      if buffer :
        cells.append( new_code_cell( ''.join( buffer ) ) )
        buffer = []
    else:
      buffer.append( line )
  
  if buffer:
    cells.append( new_code_cell( ''.join( buffer ) ) )
  
  notebook = new_notebook( cells = cells )
  
  if not ipynb_path:
    ipynb_path = py_path.replace( '.py', '.ipynb' )
  
  with open( ipynb_path, 'w', encoding = 'utf-8' ) as f :
    nbformat.write( notebook, f )
  
  print(f"✅ Converted: { py_path } → { ipynb_path }")

def batch_convert_py_to_ipynb( directory ) :
  
  if not os.path.isdir( directory ) :
    raise NotADirectoryError( f"Not a valid directory: {directory}" )
  
  py_files = [ f for f in os.listdir( directory ) if f.endswith( '.py' ) ]
  
  if not py_files :
    print( "No .py files found in the directory." )
    return
  
  for py_file in py_files :
    full_path = os.path.join( directory, py_file )
    convert_py_to_ipynb( full_path )

# 🔍 Example usage:
# Run from terminal or script:
# python batch_converter.py /path/to/your/scripts

if __name__ == '__main__' :
  if len( sys.argv ) != 2:
    print( "Usage: python batch_converter.py <directory_path>" )
  else:
    batch_convert_py_to_ipynb( sys.argv[1] )

```
